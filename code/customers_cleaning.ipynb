```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('C:\\Data Analyst Projects\\Customer Churn & Retention Analysis\\customers.csv')
```

```python
df_customers = df.copy()
```

```python
df_customers = df_customers.drop_duplicates()
df_customers = df_customers.reset_index(drop=True)
```

```python
df_customers['signup_date'] = df_customers['signup_date'].str.strip()
df_customers['signup_date'] = pd.to_datetime(df_customers['signup_date'], format='mixed', errors='coerce')
```

```python
df_customers['last_order_date'] = df_customers['last_order_date'].str.strip()

df_customers['last_order_date'] = np.where(df_customers['last_order_date'] == '0000-00-00', np.nan, df_customers['last_order_date'])
df_customers['last_order_date'] = np.where(df_customers['last_order_date'] == 'N/A', np.nan, df_customers['last_order_date'])
df_customers['last_order_date'] = np.where(df_customers['last_order_date'] == 'not available', np.nan, df_customers['last_order_date'])
df_customers['last_order_date'] = np.where(df_customers['last_order_date'] == 'NULL', np.nan, df_customers['last_order_date'])

df_customers['last_order_date'] = pd.to_datetime(df_customers['last_order_date'], format='mixed', errors='coerce')
```

```python
df_customers['age'] = df_customers['age'].str.strip()

df_customers['age'] = np.where(df_customers['age'] == '-', np.nan, df_customers['age'])
df_customers['age'] = np.where(df_customers['age'] == '?', np.nan, df_customers['age'])
df_customers['age'] = np.where(df_customers['age'] == 'N/A', np.nan, df_customers['age'])
df_customers['age'] = np.where(df_customers['age'] == 'NULL', np.nan, df_customers['age'])
df_customers['age'] = np.where(df_customers['age'] == 'unknown', np.nan, df_customers['age'])
```

```python
df_customers['age'] = pd.to_numeric(df_customers['age'], errors='coerce').astype('Int64')

age_groups = []

for age in df_customers['age']:
    if pd.isna(age):
        age_groups.append(pd.NA)
    elif age < 31:
        age_groups.append('Adolescent')
    elif age >= 55:
        age_groups.append('Senior')
    else:
        age_groups.append('Middle Age')

df_customers['age_group'] = age_groups

target_col = df_customers.pop('age_group')
df_customers.insert(4, 'age_group', target_col)
```

```python
df_customers['gender'] = df_customers['gender'].str.strip()
df_customers['gender'] = df_customers['gender'].str.upper()

df_customers['gender'] = np.where(df_customers['gender'] == 'N/A', np.nan, df_customers['gender'])

df_customers['gender'] = np.where(df_customers['gender'] == 'F', 'Female', df_customers['gender'])
df_customers['gender'] = np.where(df_customers['gender'] == 'FEMAIL', 'Female', df_customers['gender'])
df_customers['gender'] = np.where(df_customers['gender'] == 'FEMALE', 'Female', df_customers['gender'])

df_customers['gender'] = np.where(df_customers['gender'] == 'M', 'Male', df_customers['gender'])
df_customers['gender'] = np.where(df_customers['gender'] == 'MALE', 'Male', df_customers['gender'])

df_customers['gender'] = np.where(df_customers['gender'] == 'NON-BINARY', 'Non-Binary', df_customers['gender'])

df_customers['gender'] = np.where(df_customers['gender'] == 'O', 'Other', df_customers['gender'])
df_customers['gender'] = np.where(df_customers['gender'] == 'OTHER', 'Other', df_customers['gender'])

df_customers['gender'] = np.where(df_customers['gender'] == 'PREFER NOT TO SAY', 'Prefer Not to Say', df_customers['gender'])

df_customers['gender'] = np.where(df_customers['gender'] == 'U', 'Unspecified', df_customers['gender'])
df_customers['gender'] = np.where(df_customers['gender'] == 'UNSPECIFIED', 'Unspecified', df_customers['gender'])
```
